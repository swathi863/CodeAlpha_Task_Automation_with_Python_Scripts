```python
# Upload files
from google.colab import files
import os
import shutil

uploaded = files.upload()

# File type categories
file_types = {
    "Images": [".jpg", ".jpeg", ".png", ".gif"],
    "Documents": [".pdf", ".docx", ".txt", ".xlsx", ".csv"],
    "Others": []
}

os.makedirs("organized", exist_ok=True)

# Organize files
for file in uploaded:
    ext = os.path.splitext(file)[1].lower()
    moved = False
    for folder, extensions in file_types.items():
        if ext in extensions:
            folder_path = os.path.join("organized", folder)
            os.makedirs(folder_path, exist_ok=True)
            shutil.move(file, os.path.join(folder_path, file))
            moved = True
            break
    if not moved:
        other_path = os.path.join("organized", "Others")
        os.makedirs(other_path, exist_ok=True)
        shutil.move(file, os.path.join(other_path, file))

print("✅ Files organized into the 'organized' folder.")
```

```python
import pandas as pd

uploaded = files.upload()
filename = next(iter(uploaded))

# Load and clean CSV
df = pd.read_csv(filename)
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].str.strip()
df.dropna(how='all', inplace=True)

# Save and download
cleaned_file = "cleaned_" + filename
df.to_csv(cleaned_file, index=False)
files.download(cleaned_file)
print("✅ Cleaned CSV downloaded.")
```

```python
import time

def delete_old_files(folder=".", days_old=1):
    cutoff = time.time() - (days_old * 86400)
    deleted = 0

    for f in os.listdir(folder):
        path = os.path.join(folder, f)
        if os.path.isfile(path) and os.path.getmtime(path) < cutoff:
            os.remove(path)
            print(f"🗑 Deleted: {f}")
            deleted += 1
    if deleted == 0:
        print("✅ No old files found.")
    else:
        print(f"✅ Deleted {deleted} old files.")

# Example: delete files older than 1 day
delete_old_files(days_old=1)
```
